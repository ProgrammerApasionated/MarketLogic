```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


profiles = pd.read_csv("../data/profiles_38_52.csv")
responses = pd.read_csv("../data/responses_172_244.csv")
sessions = pd.read_csv("../data/sessions_56_75.csv")
profiles["categoria"] = profiles["perfil_id"].str.split("_").str[0]
# Visualizamos la distribución de categorías detectadas.
# Un gráfico de barras permite ver de forma clara qué perfiles aparecen más.
sns.countplot(data=profiles, x="categoria", palette="Set2")
plt.title("Distribución de perfiles encontrados")
plt.show()

# Calculamos cuántas respuestas tiene cada sesión.
responses_per_session = responses.groupby("session_id").size().reset_index(name="num_respuestas")

# Visualizamos la distribución del número de respuestas por sesión.
sns.histplot(responses_per_session["num_respuestas"],bins=10)
plt.title("Número de respuestas por sesión")
plt.show()
```
